```python
import pandas as pd
import numpy as np
```

```python
mat_1 = np.array([[4, 3], [1, 2]])
```

```python
mat_1
```

```python
from sklearn.decomposition import NMF
```

```python
model = NMF(n_components=2, 
            init='random', 
            random_state=0)
W = model.fit_transform(mat_1)
```

```python
W.shape
```

```python
W
```

```python
H = model.components_
```

```python
H
```

```python
np.round(W@H, 2)
```
